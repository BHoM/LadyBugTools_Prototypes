```python
import pandas as pd
import numpy as np
from pathlib import Path

# load the default systems
pth = Path("./").absolute()/ "masterplanenergydemand" / "data" / "default_systems.csv"
df = pd.read_csv(pth, header=0)

# copy one set of systems to another new set of systems
old_system_builting_type = "COMMERCIAL_STRIP_MALL"
new_system_building_type = "EXHIBITION"

# check it does not already exist
if new_system_building_type in df["building_type"].values:
    raise ValueError(f"{new_system_building_type} already exists in the set of default systems ... are you sure you want to add more?")

# copy the systems to the new building type
df = pd.concat([
    df,
    pd.DataFrame([[new_system_building_type] + i for i in df[df["building_type"] == old_system_builting_type].iloc[:, 1:].values.tolist()], columns=df.columns),
], axis=0).reset_index(drop=True)

# overwrite the file
df.to_csv(pth, index=False)
```
